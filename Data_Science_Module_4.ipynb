Basic list analysis
Question: Given the following expenses, calculate the total, average, highest expense, lowest expense and number of expenses.

```python
expenses = [1200, 2500, 800, 3000, 1500, 4000]

# Calculate total
total = sum(expenses)

# Calculate average
average = total / len(expenses)

# Highest expense
highest = max(expenses)

# Lowest expense
lowest = min(expenses)

# Number of expenses
count = len(expenses)

print("Total expenses:", total)
print("Average expenses:", average)
print("Highest expense:", highest)
print("Lowest expense:", lowest)
print("Number of expenses:", count)
```

Employee performance classification

Create a function called classify_marks() that classifies marks according to the following conditions:
- 80 and above: Excellent
- 60–79: Good
- 50–59: Average
- Below 50: Fail

```python
def classify_marks(marks):

    if marks >= 80:
        return "Excellent"

    elif marks >= 60:
        return "Good"

    elif marks >= 50:
        return "Average"

    else:
        return "Fail"


print(classify_marks(85))
print(classify_marks(70))
print(classify_marks(55))
print(classify_marks(40))
```

Integrated employee salary analysis

```python
employees = {
    "John": 35000,
    "Mary": 45000,
    "Peter": 28000,
    "Grace": 60000,
    "James": 40000,
    "Anne": 52000,
    "David": 32000
}


# Calculate total salaries
def calculate_total_salary(employees):
    return sum(employees.values())


# Calculate average salary
def calculate_average_salary(employees):
    return sum(employees.values()) / len(employees)


# Identify highest-paid employee
def highest_paid_employee(employees):
    return max(employees, key=employees.get)


# Identify lowest-paid employee
def lowest_paid_employee(employees):
    return min(employees, key=employees.get)


# Display employees earning above KSh 40,000
def high_earning_employees(employees):
    for name, salary in employees.items():
        if salary > 40000:
            print(name, salary)


# Generate complete salary report
def salary_report(employees):

    total = calculate_total_salary(employees)
    average = calculate_average_salary(employees)
    highest = highest_paid_employee(employees)
    lowest = lowest_paid_employee(employees)

    print("EMPLOYEE SALARY REPORT")
    print("----------------------")

    print("Total salaries:", total)
    print("Average salary:", round(average, 2))

    print("Highest-paid employee:", highest, employees[highest])

    print("Lowest-paid employee:", lowest, employees[lowest])

    print("\nEmployees earning above KSh 40,000:")
    high_earning_employees(employees)


# Call the report function
salary_report(employees)
```
